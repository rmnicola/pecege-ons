Este notebook explora uma semana da carga de energia do país, de 01 a 07/01/2025, para responder às perguntas das três personas do desafio. Ele não é o recado: são os gráficos de quem está investigando os dados. O recado você faz depois, em `aula3/recado.py`.

Fonte: ONS, [Curva de Carga Horária](https://dados.ons.org.br/dataset/curva-carga), extraído em 13/10/2026.

```python
import matplotlib.pyplot as plt

from fontes import dia_tipico, ler_semana, participacao_no_pico, pico_diario

semana = ler_semana()
semana.head()
```

## Os dados

Cada linha é uma hora; cada coluna, um subsistema: **SE** (Sudeste/Centro-Oeste), **S** (Sul), **NE** (Nordeste) e **N** (Norte). Os valores estão em **MWmed**: a energia consumida na hora, expressa como potência média. São 168 horas (7 dias × 24).

Dois detalhes do calendário mudam a leitura: **quarta, 01/01, é feriado**, e **04 e 05/01 são sábado e domingo**.

## Gráfico 1 · A semana hora a hora

```python
semana.plot(figsize=(12, 5), title="Carga horária");
```

- O SE tem, sozinho, cerca de três vezes a carga do NE ou do S.
- Todos os subsistemas sobem de dia e caem de madrugada: o desenho se repete a cada 24 horas.
- O primeiro dia, o feriado, é o mais baixo da semana em todos os subsistemas.

Com escalas tão diferentes, o que acontece no N e no NE quase não aparece. Para comparar o formato das curvas, olhe o dia típico.

## Gráfico 2 · O dia típico

A média de cada hora do dia na semana, por subsistema e no total (a soma dos quatro).

```python
tipico = dia_tipico(semana)
tipico.plot(figsize=(12, 5), title="Carga média por hora");
tipico.round(0)
```

- O **total chega ao pico às 20h**, com cerca de 86.800 MWmed, e ao mínimo às 5h, com cerca de 67.400.
- SE e S chegam ao pico às 20h; NE e N, mais tarde: 22h e 23h.
- Para quem opera o sistema, o fim da tarde e o começo da noite pedem mais atenção.

É uma média de sete dias, com feriado e fim de semana: ela suaviza o que acontece em cada dia.

## Gráfico 3 · O pico de cada dia

A maior carga horária de cada dia, somando os subsistemas, e a hora em que ela aconteceu.

```python
picos = pico_diario(semana)
picos["pico"].plot.bar(figsize=(12, 5), title="Pico diário");
picos.round(0)
```

- Nos dias úteis, o pico passa de 88.000 MWmed. No feriado, ficou em 77.889; no domingo, em 83.688.
- Em seis dos sete dias, o pico veio às 20h. Na terça, 07/01, veio às 14h, e foi o maior da semana: 91.501 MWmed.
- Uma semana só não diz se um pico às 14h é tendência ou exceção. Os dados também não dizem o porquê.

## Gráfico 4 · Quem pesa no pico

A participação de cada subsistema na hora de maior carga da semana (07/01, 14h).

```python
no_pico = participacao_no_pico(semana)
no_pico.plot.pie(figsize=(6, 6), title="Participação no pico");
no_pico.round(1)
```

- No pico, o **SE respondeu por 56%** da carga; S, 19%; NE, 16%; N, 9%.
- Na média da semana, o peso do SE é parecido, 55%. O que acontece no SE define o sistema.

A pizza mostra que o SE é mais da metade, mas dificulta comparar S e NE, que têm fatias parecidas.

## Gráfico 5 · Como a carga total se distribui

```python
total = semana.sum(axis=1)
total.plot.hist(figsize=(12, 5), title="Distribuição da carga");
total.describe().round(0)
```

- A carga total variou de 59.826 MWmed (feriado, 8h) a 91.501 MWmed (terça, 14h). Metade das horas ficou abaixo de 76.903 MWmed.
- As cinco horas mais baixas da semana são todas da manhã do feriado.
